```python
import pandas as pd
import numpy as np

y_train = pd.read_csv("../dataset/processed/y_train.csv").to_numpy().ravel()
y_test = pd.read_csv("../dataset/processed/y_test.csv").to_numpy().ravel()
X_train = pd.read_csv("../dataset/processed/X_train.csv").to_numpy()
X_test = pd.read_csv("../dataset/processed/X_test.csv").to_numpy()
```

```python
print(y_train.shape, y_train.dtype)
print(y_test.shape, y_test.dtype)
print(X_train.shape, X_train.dtype)
print(X_test.shape, X_test.dtype)
```

```python
m , n = X_train.shape

print(m, n)

w = np.zeros(n)
b = 0.0

epochs = 5000
rate = 0.01
```

```python
# Predicting the values

def predict( w, x, b ):

  y_pred = (x @ w) + b

  return y_pred

y_pred = predict(w, X_train, b)
print(y_pred[:5], y_pred.shape)
```

```python
# Calculating the loss function(mse)

def compute_cost(y, X, w, b):
    y_pred = predict( w, X, b)
    cost = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)
    return cost

initial_cost = compute_cost(y_train, X_train, w, b)
print(round(initial_cost, 4))
```

```python
# Calculating the gradients

def compute_gradient( y, X, w, b):
    y_pred = predict( w, X, b)
    err = y_pred - y

    dw = (1 / m) * (X.T @ err)
    db = (1 / m) * np.sum(err)

    return dw, db

dw, db = compute_gradient( y_train, X_train, w, b)
print(db, dw.shape)
```

```python
# Training the model and updating weights using epochs and learning rate

def train( y, X, w, b, rate, epochs):

    cost_history = []
    
    for i in range(epochs):
        dw, db = compute_gradient( y, X, w, b)
        w = w - (rate * dw)
        b = b - (rate * db)
        cost_history.append(compute_cost( y, X, w, b))
        

    return w, b, cost_history

w = np.zeros(n)
b = 0.0

w, b, cost_history = train( y_train, X_train, w, b, rate, epochs)

print(w, round(b, 4))
print(round(cost_history[0], 4), round(cost_history[500], 4), round(cost_history[999], 4))

```

```python
print(b)

y_test_pred = predict(w, X_test, b)

print("\n=============== TESTING ===============\n")
print("ACTUAL: ", y_test[:5])
print("PREDIC: ", y_test_pred[:5])
```

```python
mae = np.mean(np.abs(y_test_pred - y_test))
rmse = np.sqrt(np.mean((y_test_pred - y_test) ** 2))

SS_res = np.sum((y_test - y_test_pred) ** 2)
SS_tot = np.sum((y_test - np.mean(y_test)) ** 2)
r2 = 1 - (SS_res / SS_tot)

print("\n=============== EVALUATION ===============\n")
print("MAE:  ", mae)
print("RMSE: ", rmse)
print("R^2:  ", r2)
```

```python
import joblib

scratch_bundle = {
    "weights": w,
    "bias": b,
    "cost_history": cost_history
}
joblib.dump(scratch_bundle, "../model/scratch_model.joblib")
```
