```python
from sklearn.linear_model import LassoCV
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import pandas as pd
import numpy as np

y_train = pd.read_csv("../../dataset/processed/endterm/y_train.csv")['G3']
y_test = pd.read_csv("../../dataset/processed/endterm/y_test.csv")['G3']
X_train = pd.read_csv("../../dataset/processed/endterm/X_train.csv")
X_test = pd.read_csv("../../dataset/processed/endterm/X_test.csv")
```

```python
model = LassoCV(cv=5, random_state=42)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
print(np.sum(model.coef_ == 0))
print(model.alpha_)

mae = mean_absolute_error(y_test, predictions)
rmse = root_mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print(round(mae, 4))
print(round(rmse, 4))
print(round(r2, 4))
```

```python
import joblib

model_bundle = {
    "model": model,
    "feature_names": list(X_train.columns)
}
joblib.dump(model_bundle, "../../model/student_model_endterm_lasso.joblib")
```
