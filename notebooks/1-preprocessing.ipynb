```python
import pandas as pd
import numpy as np

df = pd.read_csv("../dataset/raw/student-por.csv", sep=';')
```

```python
y = df['G3']
X = df.drop(columns=['G1', 'G2', 'G3'])

print(y.shape)
print(X.shape)
```

```python
X_encoded = pd.get_dummies(X, drop_first = True)
print(X_encoded.shape)

# One-Hot encoding for multivalued columns while dropping duplicate columns
```

```python
# Splitting the training and test split, while shuffling using random state

shuffled_X = X_encoded.sample(frac = 1, random_state= 42).reset_index(drop=True)
shuffled_y = y.sample(frac = 1, random_state= 42).reset_index(drop=True)

split = int(len(shuffled_X) * 0.8)

X_train = shuffled_X.iloc[ : split]
X_test = shuffled_X.iloc[split : ]

y_train = shuffled_y.iloc[ : split]
y_test = shuffled_y.iloc[split : ]

print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
```

```python
# Standardization using Z-score

means = X_train.mean()
stds = X_train.std()

means.to_csv("../dataset/processed/means.csv")
stds.to_csv("../dataset/processed/stds.csv")

X_train_scaled = (X_train - means) / stds 
X_test_scaled = (X_test - means) / stds

print(means.shape, stds.shape)

print(X_test_scaled['absences'].mean(), X_train_scaled['absences'].std())
```

```python
X_train_scaled.to_csv("../dataset/processed/X_train.csv", index=False)
X_test_scaled.to_csv("../dataset/processed/X_test.csv", index=False)
y_train.to_csv("../dataset/processed/y_train.csv", index=False)
y_test.to_csv("../dataset/processed/y_test.csv", index=False)
```
