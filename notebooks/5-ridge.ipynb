```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
import pandas as pd
import numpy as np

y_train = pd.read_csv("../dataset/processed/y_train.csv")['G3']
y_test = pd.read_csv("../dataset/processed/y_test.csv")['G3']
X_train = pd.read_csv("../dataset/processed/X_train.csv")
X_test = pd.read_csv("../dataset/processed/X_test.csv")
```

```python
model = Ridge(alpha=0.1)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
print(np.sum(model.coef_ == 0))

mae = mean_absolute_error(y_test, predictions)
rmse = root_mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print(round(mae, 4))
print(round(rmse, 4))
print(round(r2, 4))
```

```python
from sklearn.linear_model import RidgeCV

model2 = RidgeCV(alphas=[0.01, 0.1, 1.0, 100.0, 149.0, 150.0, 155.0, 175.0], cv=5)
model2.fit(X_train, y_train)

predictions2 = model2.predict(X_test)
print(np.sum(model2.coef_ == 0))
print(model2.alpha_)

mae = mean_absolute_error(y_test, predictions2)
rmse = root_mean_squared_error(y_test, predictions2)
r2 = r2_score(y_test, predictions2)


print(round(mae, 4))
print(round(rmse, 4))
print(round(r2, 4))
```
